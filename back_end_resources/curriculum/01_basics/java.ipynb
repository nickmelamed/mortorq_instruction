## A Notebook Cell Is Not a Real `.java` File

JShell (the kernel behind this notebook) is a Java **REPL** — it lets you run top-level statements and declarations directly, no wrapper required. A real `.java` file doesn't get that convenience: every one needs at least one class, and an entry point with the exact signature `public static void main(String[] args)` (see `comparison.md`) — the JVM refuses to run anything without it.

You won't have to write that boilerplate yourself on the robot: every WPILib project already has a `Main.java` with `main` set up for you, and command-based robot code just fills in `Robot.java` and friends. But keep the `public static void main` requirement in mind for the moment you write or read a standalone `.java` file outside a notebook or a WPILib project — that's the one piece of ceremony this notebook is hiding from you.

## Variables and Types

Java is **statically typed**: every variable's type is declared up front and checked before the code will even compile. You cannot assign a `String` to a variable declared `int` — the compiler rejects it immediately, rather than waiting to find out at runtime.

```java
// Scoring constants. `final` means this variable can be assigned exactly once —
// Java's actual enforced version of "this shouldn't change" (unlike Python's naming-only convention).
final int POINTS_LOW_GOAL = 2;
final int POINTS_HIGH_GOAL = 5;
final int POINTS_CLIMB = 10;

// Match totals for one robot
int lowGoalCount = 6;
int highGoalCount = 3;
boolean climbed = true;

System.out.println(lowGoalCount + ", " + highGoalCount + ", " + climbed);
```

Every declaration above states its type explicitly: `int`, `boolean`. Try changing `int lowGoalCount = 6;` to `int lowGoalCount = "six";` in a scratch cell — it won't run. That check happening before the program executes is exactly the static-typing idea from `00_why_three_languages`.

## Functions

Java calls these **methods**, and every one declares its return type and each parameter's type explicitly. `calculateScore` below returns an `int` and takes three typed parameters.

```java
int calculateScore(int lowGoalCount, int highGoalCount, boolean climbed) {
    int total = lowGoalCount * POINTS_LOW_GOAL + highGoalCount * POINTS_HIGH_GOAL;
    if (climbed) {
        total += POINTS_CLIMB;
    }
    return total;
}

int score = calculateScore(lowGoalCount, highGoalCount, climbed);
System.out.println("Total score: " + score);
```

Note the method signature: `int calculateScore(int, int, boolean)`. Java (and C++, later) require you to spell this out — the compiler uses it to check every call site matches before the program can even run.

## Control Flow

Same rangefinder scenario as the Python notebook: a sequence of distance readings as an arm swings toward a hard stop, and a decision per reading about whether to stop.

```java
final double STOP_DISTANCE_INCHES = 6.0;

boolean shouldStopArm(double distanceInches) {
    return distanceInches <= STOP_DISTANCE_INCHES;
}

// Simulated rangefinder readings taken once per control loop iteration
double[] readings = {24.0, 18.0, 12.0, 9.0, 6.5, 5.0, 3.0};

for (double distance : readings) {
    if (shouldStopArm(distance)) {
        System.out.println(String.format("%5.1f in -> STOP", distance));
    } else {
        System.out.println(String.format("%5.1f in -> continue", distance));
    }
}
```

`double[] readings` is a Java **array**: a fixed-size, single-typed collection (every element must be a `double`). The `for (double distance : readings)` loop is Java's "enhanced for" syntax — it walks the array by value, same idea as Python's `for distance in readings:`. We'll cover arrays and other collections properly in `03_data_structures_algorithms`.

The `for` loop above always checks every reading in the list. A **while loop** instead keeps going only as long as a condition holds, and stops the moment it doesn't — closer to how the arm actually behaves: you take one reading at a time, and you don't know ahead of time how many it'll take before one says stop.

```java
int i = 0;
while (i < readings.length && !shouldStopArm(readings[i])) {
    System.out.println(String.format("%5.1f in -> continue", readings[i]));
    i++;
}
if (i < readings.length) {
    System.out.println(String.format("%5.1f in -> STOP", readings[i]));
}
```

Notice the last reading (3.0) is never checked — the loop already stopped at 5.0. That's the real difference from the `for` loop above: a `for` loop marches through the whole collection regardless, a `while` loop can bail out the instant its condition changes.
