## Variables and Types

Python variables don't need a declared type — you just assign a value, and Python figures out the type from the value itself. This is **dynamic typing**: the same name could hold an `int` now and a `str` later (though doing that on purpose is usually a bad idea — it makes code confusing to read).

```python
# Scoring constants. ALL_CAPS is a Python convention for "this shouldn't change" —
# Python has no true constant keyword, this is a naming convention only.
POINTS_LOW_GOAL = 2
POINTS_HIGH_GOAL = 5
POINTS_CLIMB = 10

# Match totals for one robot
low_goal_count = 6      # int
high_goal_count = 3     # int
climbed = True          # bool

print(type(low_goal_count), type(climbed))
```

Notice we never wrote `int` or `bool` anywhere — Python inferred both from the values on the right-hand side of `=`. You can check a variable's current type at any time with the built-in `type()` function, which is exactly what we just did.

## Functions

Rather than compute the score inline, we wrap the logic in a function. Python function parameters don't require declared types either (though you *can* add optional type hints — more on those in `04b_python_intricacies`).

```python
def calculate_score(low_goal_count, high_goal_count, climbed):
    """Return total match points for one robot."""
    total = low_goal_count * POINTS_LOW_GOAL + high_goal_count * POINTS_HIGH_GOAL
    if climbed:
        total += POINTS_CLIMB
    return total

score = calculate_score(low_goal_count, high_goal_count, climbed)
print(f"Total score: {score}")
```

`calculate_score` takes three parameters and returns one value — same shape a function would have in any language. The body just happens to look shorter in Python because there's no type annotations or curly braces required.

## Control Flow

Now the second example: a rangefinder on an arm reports a sequence of distance readings (in inches) as the arm swings toward a hard stop. We want to stop the arm once it gets within 6 inches of the stop. This needs a loop (to process each reading) and a conditional (to decide, per reading, whether to stop).

```python
STOP_DISTANCE_INCHES = 6.0

def should_stop_arm(distance_inches):
    """True once the arm is close enough to the hard stop that it should halt."""
    return distance_inches <= STOP_DISTANCE_INCHES

# Simulated rangefinder readings taken once per control loop iteration
readings = [24.0, 18.0, 12.0, 9.0, 6.5, 5.0, 3.0]

for distance in readings:
    if should_stop_arm(distance):
        print(f"{distance:5.1f} in -> STOP")
    else:
        print(f"{distance:5.1f} in -> continue")
```

The `for distance in readings:` loop is Python's standard way to iterate directly over the *values* in a collection — no manual index bookkeeping required. Each pass through the loop is one "tick" of what would, on a real robot, be the periodic control loop running many times a second.

The `for` loop above always checks every reading in the list. A **while loop** instead keeps going only as long as a condition holds, and stops the moment it doesn't — closer to how the arm actually behaves: you take one reading at a time, and you don't know ahead of time how many it'll take before one says stop.

```python
i = 0
while i < len(readings) and not should_stop_arm(readings[i]):
    print(f"{readings[i]:5.1f} in -> continue")
    i += 1
if i < len(readings):
    print(f"{readings[i]:5.1f} in -> STOP")
```

Notice the last reading (3.0) is never checked — the loop already stopped at 5.0. That's the real difference from the `for` loop above: a `for` loop marches through the whole collection regardless, a `while` loop can bail out the instant its condition changes.
